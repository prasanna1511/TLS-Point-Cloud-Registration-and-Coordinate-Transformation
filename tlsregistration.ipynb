```python
import numpy as np

def polar2cart(d, theta, phi):
   x = d * np.sin(theta) * np.sin(phi) 
   y = d * np.sin(theta) * np.cos(phi)
   z = d * np.cos(theta)
   return x, y, z
```

```python
s1  =np.loadtxt('S1_PointCloud.txt')
print(s1.shape)

# print column 1 ,then 3 and 2 column with all rows
s_cart = polar2cart(s1[:,0], s1[:,2], s1[:,1])
s_cart = np.array(s_cart).T

print(s_cart.shape)
#save as txt file
np.savetxt('S1_cart.txt', s_cart, fmt='%f', delimiter=' ')
```

Normalization of Axis-angle representation and compute the Rotation Matrix elements using the Rodrigues formula.


 # (θ,r) =  [r1, r2, r3].T,   ||r||=1 

```python
def transform(axes):
    a = axes[0]**2 + axes[2]**2 + axes[1]**2
    return a

print("9_10 :",  transform(rotation_9_10))
print("10_1 :",  transform(rotation_10_1))
print("1_2 :",  transform(rotation_1_2))
print("2_3 :",  transform(rotation_2_3))
print("3_4 :",  transform(rotation_3_4))
print("4_W :",  transform(rotation_4_W))
print("the normalized rotation axis has a length of 1.0 as shown above which are normalized around 1")
```

# Axis-Angle to Rotation Matrix

Rodrigues Formula

R(r,θ) = I3 + sinθ Sr + (1- cosθ) *Sr*Sr


θ

```python
def axis_2_rot(axis, angle):
    axis = axis/np.sqrt(transform(axis))
    cos = np.cos(angle)
    sin = np.sin(angle)
    trans = 1 - cos
    x = axis[0]
    y = axis[1]
    z = axis[2]

    R =  [[trans*x**2 +cos ,  trans*x*y -z*sin,  trans*x*z + y*sin],
             [trans*x*y +z*sin,  trans*y**2 +cos,  trans*y*z - x*sin],
             [trans*x*z - y*sin , trans*y*z + x*sin , trans*z**2 + cos]]
    return R
```

```python
def register(path1, path2, r, t, georef=False):
    
    pointCloud1 = np.loadtxt(path1)
    
    #print(pointCloud1)
    # Rotation Matrix
    R = axis_2_rot(r[0:3], r[3])
    # Create a new point cloud file with name in the fashion of path1_path2.txt
    if georef == False:
        path1_parts = path1.split("/")
        path2_parts = path2.split("/")
    
    # Check if the paths have the expected structure
    if len(path1_parts) >= 3 and len(path2_parts) >= 3:
        file_name1 = path1_parts[-1].split(".")[0]
        file_name2 = path2_parts[-1].split(".")[0]
        path1_path2 = path1.split("/")[0] + "/" +    path1.split("/")[1] + "/" + path1.split("/")[2].split(".")[0] +  "_" + path2.split("/")[2].split(".")[0] + ".txt"
    else:
        path1_path2 = "GeoRef.txt"
    #print(path1_path2)
    pointCloud1_in_pointCloud2 = open(path1_path2, "w")
    for line in pointCloud1:
        x = line[0]
        y = line[1]
        z = line[2]
        p = np.array([x, y, z])
        p_rotated = np.matmul(R, np.transpose(p))
        p_trasformed = p_rotated + t # Adding Translation
        # Adding p_transformed to path1_path2.txt
        pointCloud1_in_pointCloud2.write("   " + str(p_trasformed[0]) + "   " + str(p_trasformed[1]) + "   " + str(p_trasformed[2]) + "\n")
    pointCloud1_in_pointCloud2.close()

    if georef == False:
        pointCloud2 = np.loadtxt(path2)
        # Combine pointCloud2 into path1_path2.txt
        pointCloud1_in_pointCloud2 = open(path1_path2, "a")
        for line in pointCloud2:
            x = line[0]
            y = line[1]
            z = line[2]
            pointCloud1_in_pointCloud2.write("   " + str(x) + "   " + str(y) + "   " + str(z) + "\n")
        pointCloud1_in_pointCloud2.close()

# Registering S9 to S10
register("S9_cart1.txt", "S10_cart.txt", rotation_9_10, translation_9_10)

# Registering S10 to S1
register("S9_cart1_S10_cart.txt", "S1_cart.txt", rotation_10_1, translation_10_1)
# register("S9_cart1_S10_cart_S1_cart.txt", "S2_cart.txt", rotation_1_2, translation_1_2)
# register("S9_cart1_S10_cart_S1_cart_S2_cart.txt", "S3_cart.txt", rotation_2_3, translation_2_3)
# register("S9_cart1_S10_cart_S1_cart_S2_cart_S3_cart.txt", "S4_cart.txt", rotation_3_4, translation_3_4)
# register("S9_cart1_S10_cart_S1_cart_S2_cart_S3_cart_S4_cart.txt", "GeoRef.txt", rotation_4_W, translation_4_W, georef=True)
```
